```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([0,50])
y=np.array([0,180])

plt.plot(x,y)
plt.title("Draw a line")
plt.xlabel("x-axis")
plt.ylabel("y-aix")

plt.show()
```

```python
x=np.array([0,2,3])
y=np.array([2,4,0])

plt.plot(x,y)
plt.title("sample graph")
plt.xlabel("x-axis")
plt.ylabel("y-aix")

plt.show()
```

```python
x1=np.array([10,20,30])
y1=np.array([20,40,10])

x2=np.array([10,20,30])
y2=np.array([40,10,30])
plt.plot(x1,y1,label="line1")
plt.plot(x2,y2,label="line2")
plt.legend()
plt.xlabel("x-axis")
plt.show()
```

```python
x1=np.array([10,20,30])
y1=np.array([20,40,10])

x2=np.array([10,20,30])
y2=np.array([40,10,30])
plt.plot(x1,y1,label="line1",color="blue")
plt.plot(x2,y2,label="line2",color="red",linewidth="7")
plt.legend()
plt.xlabel("x-axis")
plt.show()
```

```python
x1=np.array([10,20,30,60])
y1=np.array([20,40,10,20])
plt.plot(x1,y1,linestyle="dotted",color="red",marker='o')
plt.xlabel("x-axis")
plt.show()
```
